```python
import pandas as pd

datos = {'a': 10, 'b': 20, 'c': 30}

serie = pd.Series(datos)
print(serie)
```

```python
notas = {
    'Matemáticas': 4.5,
    'Inglés': 3.8,
    'Programación': 5.0
}

serie_notas = pd.Series(notas)
print(serie_notas)
```

```python
temperaturas = {
    'Lunes': 30,
    'Martes': 28,
    'Miércoles': 27
}

serie_temp = pd.Series(temperaturas)

print(serie_temp)
print("Índices:", serie_temp.index)
```

```python
edades = {
    'Ana': 20,
    'Luis': 22,
    'María': 19
}

serie_edades = pd.Series(edades)

print(serie_edades)
print("Edad máxima:", serie_edades.max())
```

```python
edades['Carlos'] = 23

serie_edades_actualizada = pd.Series(edades)

print(serie_edades_actualizada)
```
