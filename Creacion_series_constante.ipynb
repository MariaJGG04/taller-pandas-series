```python
import pandas as pd

serie_constante = pd.Series([100] * 5)

print(serie_constante)
```

```python
texto = pd.Series(["Aprendiendo Pandas"] * 8)

print(texto)
```

```python
valores = pd.Series([True] * 10)

print(valores)
```

```python
pi = pd.Series([3.14] * 5, dtype='float64')

print(pi)
print("Tipo de dato:", pi.dtype)
```

```python
# Serie numérica
numeros = pd.Series([1, 2, 3, 4, 5])

# Serie constante
constante = pd.Series([10] * 5)

# Suma de ambas
resultado = numeros + constante

print("Serie numérica:")
print(numeros)

print("Serie constante:")
print(constante)

print("Resultado:")
print(resultado)
```
