```python
import pandas as pd
import numpy as np

serie_mixta = pd.Series([1, "Hola", np.nan, 5])

print(serie_mixta)
```

```python
print("Tipo de dato:", serie_mixta.dtype)
```

```python
serie = pd.Series([1, 'dos', 3.0, True])

print(serie)
print("Tipo de dato:", serie.dtype)
```

```python
serie_variada = pd.Series([
    "2024-01-01",
    100,
    "Texto",
    3.14
])

print(serie_variada)
```

```python
serie_str = serie_mixta.astype(str)

print(serie_str)
print("Tipo de dato:", serie_str.dtype)
```
