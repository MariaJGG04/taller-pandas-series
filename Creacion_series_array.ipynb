```python
import pandas as pd
import numpy as np

# Crear array de 10 números aleatorios entre 0 y 100
array = np.random.randint(0, 101, 10)

serie = pd.Series(array)

print("Array:", array)
print("Serie:")
print(serie)
```

```python
array = np.arange(1, 11)

serie = pd.Series(array)

print(serie)
```

```python
array = np.array([1.1, 2.2, 3.3, 4.4])

serie = pd.Series(array)

print(serie)
```

```python
celsius = np.array([20, 25, 30, 35])

serie_c = pd.Series(celsius)

# Convertir a Fahrenheit
fahrenheit = (serie_c * 9/5) + 32

print("Celsius:")
print(serie_c)

print("Fahrenheit:")
print(fahrenheit)
```

```python
array = np.array([1, 2, 3, 4, 5])

serie = pd.Series(array)

resultado = serie * 10

print("Serie original:")
print(serie)

print("Serie multiplicada por 10:")
print(resultado)
```
