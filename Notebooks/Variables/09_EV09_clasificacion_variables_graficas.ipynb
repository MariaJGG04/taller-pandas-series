## Clasificacion de variables y seleccion de graficas

Este notebook identifica las variables del dataset, propone su clasificacion estadistica y recomienda la visualizacion mas conveniente para cada una.

```python
%matplotlib inline

from pathlib import Path
from IPython.display import Markdown, display

import pandas as pd

ROOT = Path.cwd().resolve()
ARTIFACTS_DIR = ROOT / "artifacts" / "ev09"
ARTIFACTS_DIR.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(ROOT / "data" / "raw" / "dataset.csv")
df["Database"] = df["Database"].fillna("Sin asignar")

display(Markdown("### Vista previa del dataset"))
display(df)
```

## 1. Identificacion de variables

Se genera una tabla de reconocimiento con nombre, tipo de dato en pandas y cantidad de valores unicos.

```python
overview_df = pd.DataFrame({
    "Variable": df.columns,
    "Tipo_pandas": [str(df[col].dtype) for col in df.columns],
    "Valores_unicos": [df[col].nunique(dropna=True) for col in df.columns],
    "Ejemplo": [df[col].dropna().astype(str).iloc[0] for col in df.columns],
})
display(overview_df)
```

## 2. Tabla de clasificacion

La clasificacion se construye combinando criterio estadistico y contexto del negocio.

```python
classification = [
    {"Variable": "Job_Title", "Tipo": "Cualitativa", "Subtipo": "Nominal", "Justificacion": "Representa nombres de cargos sin orden inherente."},
    {"Variable": "Level", "Tipo": "Cualitativa", "Subtipo": "Ordinal", "Justificacion": "Expresa jerarquia laboral con orden natural."},
    {"Variable": "Location", "Tipo": "Cualitativa", "Subtipo": "Nominal", "Justificacion": "Describe la ciudad o ubicacion sin escala numerica."},
    {"Variable": "Technology", "Tipo": "Cualitativa", "Subtipo": "Nominal", "Justificacion": "Identifica tecnologias diferentes sin orden."},
    {"Variable": "Framework", "Tipo": "Cualitativa", "Subtipo": "Nominal", "Justificacion": "Nombra herramientas o frameworks sin jerarquia cuantitativa."},
    {"Variable": "Database", "Tipo": "Cualitativa", "Subtipo": "Nominal", "Justificacion": "Corresponde a motores de base de datos categorizables por nombre."},
    {"Variable": "Experience_Years", "Tipo": "Cuantitativa", "Subtipo": "Discreta", "Justificacion": "Cuenta anios de experiencia como valores enteros."},
]

classification_df = pd.DataFrame(classification)
display(classification_df)
classification_df.to_csv(ARTIFACTS_DIR / "tabla_clasificacion_variables.csv", index=False, encoding="utf-8")
```

## 3. Tabla de seleccion de graficas recomendadas

La seleccion considera el tipo de variable, la lectura deseada y el tamano del dataset.

```python
chart_selection = [
    {"Variable": "Job_Title", "Grafica_recomendada": "Barras horizontales", "Justificacion": "Permite comparar frecuencias de categorias con nombres largos."},
    {"Variable": "Level", "Grafica_recomendada": "Barras ordenadas", "Justificacion": "Respeta el orden jerarquico de la variable ordinal."},
    {"Variable": "Location", "Grafica_recomendada": "Barras agrupadas", "Justificacion": "Facilita comparar sedes o ciudades por conteo o por nivel."},
    {"Variable": "Technology", "Grafica_recomendada": "Barras", "Justificacion": "Muestra con claridad la frecuencia de cada tecnologia."},
    {"Variable": "Framework", "Grafica_recomendada": "Barras", "Justificacion": "Es adecuada para comparar categorias nominales."},
    {"Variable": "Database", "Grafica_recomendada": "Barras apiladas", "Justificacion": "Ayuda a revisar composicion por nivel u otra categoria de cruce."},
    {"Variable": "Experience_Years", "Grafica_recomendada": "Histograma y boxplot", "Justificacion": "El histograma resume distribucion y el boxplot detecta dispersion."},
]

chart_df = pd.DataFrame(chart_selection)
display(chart_df)
chart_df.to_csv(ARTIFACTS_DIR / "tabla_graficas_recomendadas.csv", index=False, encoding="utf-8")
```
