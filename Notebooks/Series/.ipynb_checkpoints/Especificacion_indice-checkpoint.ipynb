```python
import pandas as pd

personas = pd.Series(
    ['Juan', 'Ana', 'Luis'],
    index=['ID1', 'ID2', 'ID3']
)

print(personas)
```

```python
paises = pd.Series(
    ['Colombia', 'México', 'Argentina'],
    index=['CO', 'MX', 'AR']
)

print(paises)
```

```python
calificaciones = pd.Series(
    [4.5, 3.8, 4.2],
    index=['Juan', 'Ana', 'Luis']
)

print(calificaciones)
```

```python
serie = pd.Series([10, 20, 30, 40])

# Cambiar índices
serie.index = ['a', 'b', 'c', 'd']

print(serie)
```

```python
print(calificaciones['Juan'])
```
