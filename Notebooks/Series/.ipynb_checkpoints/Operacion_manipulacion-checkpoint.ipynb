```python
import pandas as pd

serie1 = pd.Series([10, 20, 30, 40, 50])
serie2 = pd.Series([5, 15, 25, 35, 45])

suma = serie1 + serie2

print("Suma:")
print(suma)
```

```python
resta = serie1 - serie2

print("Resta:")
print(resta)
```

```python
multiplicacion = serie1 * 2

print("Multiplicación por 2:")
print(multiplicacion)
```

```python
division = serie1 / 2

print("División entre 2:")
print(division)
```

```python
resultado = (serie1 + serie2) / 2

print("Promedio de ambas series:")
print(resultado)
```

```python
import pandas as pd

serie = pd.Series([1, 2, 3, 4, 5])

cuadrado = serie.apply(lambda x: x**2)

print(cuadrado)
```

```python
nombres = pd.Series(["juan", "ana", "luis"])

mayusculas = nombres.map(lambda x: x.upper())

print(mayusculas)
```

```python
import numpy as np

numeros = pd.Series([4, 9, 16, 25])

raiz = numeros.apply(np.sqrt)

print(raiz)
```

```python
def clasificar(valor):
    if valor >= 3:
        return "Aprobado"
    else:
        return "Reprobado"

notas = pd.Series([2.5, 3.5, 4.0, 2.8])

resultado = notas.apply(clasificar)

print(resultado)
```

```python
datos = pd.Series(["A", "B", "C", "A"])

reemplazo = {
    "A": "Excelente",
    "B": "Bueno",
    "C": "Regular"
}

resultado = datos.map(reemplazo)

print(resultado)
```

```python
import pandas as pd

serie = pd.Series([50, 10, 40, 20, 30])

ordenada = serie.sort_values()

print(ordenada)
```

```python
nombres = pd.Series(["Luis", "Ana", "Carlos", "Beatriz"])

ordenados = nombres.sort_values()

print(ordenados)
```

```python
serie = pd.Series(
    [10, 20, 30],
    index=['c', 'a', 'b']
)

ordenado_indice = serie.sort_index()

print(ordenado_indice)
```

```python
serie = pd.Series([10, 50, 30, 80, 60])

mayores = serie.nlargest(3)

print("3 valores más altos:")
print(mayores)
```

```python
menores = serie.nsmallest(3)

print("3 valores más bajos:")
print(menores)
```

```python
import pandas as pd

serie = pd.Series([10, 55, 70, 30, 90, 20])

mayores_50 = serie[serie > 50]

print(mayores_50)
```

```python
pares = serie[serie % 2 == 0]

print(pares)
```

```python
texto = pd.Series([
    "Aprendiendo Pandas",
    "Python es genial",
    "Pandas es poderoso",
    "Datos y análisis"
])

filtro = texto[texto.str.contains("Pandas")]

print(filtro)
```

```python
condicion = serie[(serie > 30) & (serie < 70)]

print(condicion)
```

```python
indices = serie[serie > 50].index

print("Índices de valores mayores a 50:")
print(indices)
```

```python
import pandas as pd
import numpy as np

serie = pd.Series([10, 20, np.nan, 40, np.nan])

reemplazo = serie.fillna(0)

print(reemplazo)
```

```python
sin_nulos = serie.dropna()

print(sin_nulos)
```

```python
cantidad_nulos = serie.isna().sum()

print("Cantidad de valores nulos:", cantidad_nulos)
```

```python
promedio = serie.mean()

relleno_promedio = serie.fillna(promedio)

print("Promedio:", promedio)
print(relleno_promedio)
```

```python
relleno_constante = serie.fillna(999)

print(relleno_constante)
```

```python
import pandas as pd

serie1 = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
serie2 = pd.Series([5, 15, 25], index=['b', 'c', 'd'])

resultado = serie1 + serie2

print(resultado)
```

```python
resultado = serie1.add(serie2, fill_value=0)

print(resultado)
```

```python
serie1 = pd.Series([10, None, 30], index=['a', 'b', 'c'])
serie2 = pd.Series([5, 15, 25], index=['a', 'b', 'c'])

resultado = serie1.combine_first(serie2)

print(resultado)
```

```python
serie1 = pd.Series([1, 2, 3], index=['a', 'b', 'c'])
serie2 = pd.Series([4, 5, 6], index=['b', 'c', 'd'])

print("Serie 1:")
print(serie1)

print("Serie 2:")
print(serie2)

print("Suma alineada:")
print(serie1 + serie2)
```

```python
serie1 = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
serie2 = pd.Series([1, 2, 3], index=['a', 'b', 'c'])

resultado = (serie1 * 2 + serie2) / 3

print(resultado)
```

```python
import pandas as pd

serie = pd.Series([10, 20, 30, 40], index=['a','b','c','d'])

nueva_serie = serie.reindex(['a','b','c','d','e'])

print(nueva_serie)
```

```python
serie_invertida = serie.reindex(['d','c','b','a'])

print(serie_invertida)
```

```python
serie_fill = serie.reindex(['a','b','c','d','e'], fill_value=0)

print(serie_fill)
```

```python
nueva_serie = serie.reindex(['a','b','c','d','e'])

promedio = serie.mean()

relleno = nueva_serie.fillna(promedio)

print("Promedio:", promedio)
print(relleno)
```

```python
serie_mixta = serie.reindex(['b','c','e','a'])

print(serie_mixta)
```
