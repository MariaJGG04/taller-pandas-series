```python
# shape #
import pandas as pd
data = {'Nombre': ['Ana'
,
'Luis'
,
'Marta'],

'Edad': [23, 35, 29]}

df = pd.DataFrame(data)
print(df.shape) # Salida: (3, 2)
```

```python
# size #
print(df.size) # Salida: 6
```

```python
# ndim #
print(df.ndim) # Salida: 2
```

```python
# columns #
print(df.columns) # Salida: Index(['Nombre','Edad'], dtype='object')
```

```python
# index #
print(df.index) # Salida: RangeIndex(start=0, stop=3, step=1)
```

```python
# dtypes #
print(df.dtypes)
# Salida:
# Nombre object
# Edad int64
# dtype: object
```

```python
# values #
print(df.values)
# Salida:
# [['Ana' 23]
# ['Luis' 35]
# ['Marta' 29]]
```

```python
# T #
print(df.T)
# Salida:
# 0 1 2
# Nombre Ana Luis Marta
# Edad 23 35 29
```

```python
# head #
print(df.head(2))
# Salida:
# Nombre Edad
# 0 Ana 23
# 1 Luis 35
```

```python
# tail(n) #
print(df.tail(2))
# Salida:
# Nombre Edad
# 1 Luis 35
# 2 Marta 29
```

```python
# memory_usage(deep=True) #

print(df.memory_usage(deep=True))
# Salida:
# Index 132
# Nombre 159
# Edad 24
# dtype: int64
```

```python
# empty #
print(df.empty) # Salida: False
```
