```python
# A partir de diccionarios #

import pandas as pd
data = {
'Nombre': ['Ana'
,
'Luis'
,
'Marta'],

'Edad': [23, 35, 29],
'Ciudad': ['Madrid'
,
'Barcelona'
,
'Valencia']

}
df = pd.DataFrame(data)
print(df)
```

```python
# A partir de listas #

data = [
['Ana', 23,'Madrid'],

['Luis', 35,'Barcelona'],

['Marta', 29,'Valencia']

]
df = pd.DataFrame(data, columns=['Nombre','Edad','Ciudad'])
print(df)
```

```python
# A partir de listas de diccionarios #

data = [
{'Nombre': 'Ana','Edad': 23,'Ciudad': 'Madrid'},

{'Nombre': 'Luis','Edad': 35,'Ciudad': 'Barcelona'},

{'Nombre': 'Marta','Edad': 29,'Ciudad': 'Valencia'}

]
df = pd.DataFrame(data)
print(df)
```

```python
# A partir de arrays de NumPy #

import pandas as pd
import numpy as np
data = np.array([
['Ana', 23,'Madrid'],

['Luis', 35,'Barcelona'],

['Marta', 29,'Valencia']

])
df = pd.DataFrame(data, columns=['Nombre','Edad','Ciudad'])
print(df)
```

```python
# A partir de series de pandas #

nombres = pd.Series(['Ana','Luis','Marta'])

edades = pd.Series([23, 35, 29])
ciudades = pd.Series(['Madrid','Barcelona','Valencia'])

df = pd.DataFrame({
'Nombre': nombres,
'Edad': edades,
'Ciudad': ciudades
})
print(df)
```
