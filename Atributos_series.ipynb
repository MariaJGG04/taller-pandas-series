```python
import pandas as pd
import numpy as np

serie = pd.Series([10, 20, 30, 40, 50])

print("Índice:", serie.index)
print("Valores:", serie.values)
print("Tipo de dato:", serie.dtype)
```

```python
serie = pd.Series(range(1, 21))  # 20 elementos

print("Tamaño (size):", serie.size)
print("Forma (shape):", serie.shape)
print("¿Está vacía?:", serie.empty)
```

```python
serie_dup = pd.Series([1, 2, 2, 3, 4, 4])

print("Serie:")
print(serie_dup)

print("¿Valores únicos?:", serie_dup.is_unique)
```

```python
serie_nan = pd.Series([1, 2, np.nan, 4])

print(serie_nan)

print("¿Tiene valores nulos?:", serie_nan.hasnans)
```

```python
texto = pd.Series([
    "Este es un texto largo",
    "Pandas es muy útil",
    "Aprendiendo ciencia de datos"
])

print(texto)

print("Memoria en bytes:", texto.nbytes)
```
