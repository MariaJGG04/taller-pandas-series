```python
# head(n=5) #
import pandas as pd
data = {'Nombre': ['Ana', 'Luis', 'Marta', 'Juan', 'Carlos'], 'Edad': [23, 35, 29, 40, 22]}
df = pd.DataFrame(data)
print(df.head(3))
# Salida:
# Nombre Edad
# 0 Ana 23
# 1 Luis 35
# 2 Marta 29
```

```python
# tail(n=5) #
print(df.tail(2))
# Salida:
# Nombre Edad
# 3 Juan 40
# 4 Carlos 22
```

```python
# to_numpy() #
print(df.to_numpy())
# Salida:
# [['Ana' 23]
# ['Luis' 35]
# ['Marta' 29]]
```

```python
# info() #
df.info()
# Salida:
# <class 'pandas.core.frame.DataFrame'>
# RangeIndex: 5 entries, 0 to 4
# Data columns (total 2 columns):
# # Column Non-Null Count Dtype
# --- ------ -------------- -----
# 0 Nombre 5 non-null object
# 1 Edad 5 non-null int64
# dtypes: int64(1), object(1)
# memory usage: 208.0+ bytes
```

```python
# describe() #
print(df.describe())
# Salida:
# Edad
# count 5.000000
# mean 29.800000
# std 7.726578
# min 22.000000
# 25% 23.000000
# 50% 29.000000
# 75% 35.000000
# max 40.000000
```

```python
# drop(columns) #
df_sin_edad = df.drop(columns=['Edad'])
print(df_sin_edad)
# Salida:
# Nombre
# 0 Ana
# 1 Luis
# 2 Marta
# 3 Juan
# 4 Carlos
```

```python
# dropna() #
df_con_na = pd.DataFrame({'Nombre': ['Ana','Luis', None],'Edad': [23, None, 29]})

df_sin_na = df_con_na.dropna()
print(df_sin_na)
# Salida:
# Nombre Edad
# 0 Ana 23.0
```

```python
# fillna(valor) #
df_rellenado = df_con_na.fillna(0)
print(df_rellenado)
# Salida:
# Nombre Edad
# 0 Ana 23.0
# 1 Luis 0.0
# 2 0 29.0
```

```python
# sort_values(by) #
df_ordenado = df.sort_values(by='Edad')
print(df_ordenado)
# Salida:
# Nombre Edad
# 4 Carlos 22
# 0 Ana 23
# 2 Marta 29
# 1 Luis 35
# 3 Juan 40
```

```python
# groupby(by) #
import pandas as pd
data = {'Nombre': ['Ana','Luis','Marta','Ana','Luis'],
'Edad': [23, 35, 29, 22, 36],
'Ciudad': ['Madrid','Barcelona','Valencia','Madrid','Barcelona']}

df = pd.DataFrame(data)
df_agrupado = df.groupby('Ciudad').mean(numeric_only=True)
print(df_agrupado)
# Salida:
# Edad
# Ciudad
# Barcelona 35.5
# Madrid 22.5
# Valencia 29.0
```
